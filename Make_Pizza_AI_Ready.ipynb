```python
%%writefile pizza-api.yaml
openapi: 3.0.0
info:
  title: Pizza Shop API
  version: 1.0.0
paths:
  /menu:
    get:
      summary: List available pizzas
      responses:
        '200':
          description: A list of pizzas
  /order:
    post:
      summary: Place a pizza order
      requestBody:
        content:
          application/json:
            schema:
              type: object
              properties:
                item: {type: string}
                size: {type: string}
      responses:
        '200':
          description: Order confirmation
          content:
            application/json:
              schema:
                type: object
                properties:
                  order_id: {type: string}
                  status: {type: string}
                  eta: {type: integer}
```

```python
%%bash
# 1. Generate the MCP server code from your YAML
npx openapi-mcp-generator -i pizza-api.yaml -o ./pizza-server
```

```python
%%bash
# 1. Generate the MCP server code
# If npx is not found, we install the generator locally first
npm install openapi-mcp-generator

# 2. Run the generator to transform your YAML into an MCP server
npx openapi-mcp-generator -i pizza-api.yaml -o ./pizza-server

# 3. Enter the folder and set up the server environment
cd ./pizza-server
npm install
npm run build

echo "------------------------------------------------"
echo "✅ SUCCESS: Pizza MCP Server is built and ready!"
echo "------------------------------------------------"
```

```python
import os
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI
# Note: In a real environment, you'd use the MCP Adapter
# Here is the logic for the Gemini Agent to interact with the generated server

os.environ["GOOGLE_API_KEY"] = userdata.get('GOOGLE_API_KEY')

llm = ChatGoogleGenerativeAI(model="gemini-1.5-flash")

# This simulated prompt shows how Gemini utilizes the tools we just built
user_input = "Order a large Margherita. Then add a 'Pizza Party' event to my calendar for its arrival time."

print(f"🤖 Gemini is processing: {user_input}")
print("-" * 30)
print("1. [Tool Call] pizza_server.list_menu() -> Success")
print("2. [Tool Call] pizza_server.place_order(item='Margherita', size='large') -> Order #123 (ETA: 30 mins)")
print("3. [Tool Call] calendar.create_event(title='Pizza Party', time='7:30 PM') -> Success")
print("-" * 30)
print("Final Response: Your large Margherita is ordered (Order #123)! I've also added 'Pizza Party' to your calendar for 7:30 PM.")
```
